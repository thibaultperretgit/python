```python
def longest_unique_window(values: list[int]) -> int:
    """
    Find the length of the longest contiguous subarray with all unique elements.

    Args:
        values (list[int]): A list of integers.

    Returns:
        int: The length of the longest contiguous subarray with all unique elements.
    """
    seen = set()
    left = 0
    max_length = 0

    for right in range(len(values)):
        while values[right] in seen:
            seen.remove(values[left])
            left += 1
        seen.add(values[right])
        max_length = max(max_length, right - left + 1)

    return max_length

print(longest_unique_window([]), longest_unique_window([1, 2, 3, 2, 4, 5]), longest_unique_window([7, 7, 7]))
```

```python
def longest_unique_window(values: list[int]) -> list[int]:
    """
    Find the longest contiguous subarray with all unique elements.

    Args:
        values (list[int]): A list of integers.

    Returns:
        list[int]: The longest contiguous subarray with all unique elements.
    """
    seen = set()
    left = 0
    best_start = 0
    best_length = 0

    for right in range(len(values)):
        while values[right] in seen:
            seen.remove(values[left])
            left += 1
        seen.add(values[right])
        if right - left + 1 > best_length:
            best_length = right - left + 1
            best_start = left


    return values[best_start:best_start + best_length]

print(longest_unique_window([]), longest_unique_window([1, 2, 3, 2, 4, 5]), longest_unique_window([7, 7, 7]))
```

```python
def merge_intervals(intervals: list[tuple[int, int]]) -> list[tuple[int, int]]:
    """
    Merge overlapping intervals.

    Args:
        intervals (list[tuple[int, int]]): A list of intervals represented as tuples.

    Returns:
        list[tuple[int, int]]: A list of merged intervals.
    """
    if not intervals:
        return []

    # Sort intervals based on the start time
    intervals_bis = intervals.copy()
    intervals_bis.sort(key=lambda x: x[0])
    merged = [intervals_bis[0]]

    for current in intervals_bis[1:]:
        last_merged = merged[-1]
        if current[0] <= last_merged[1]:  # Overlapping intervals
            merged[-1] = (last_merged[0], max(last_merged[1], current[1]))  # Merge
        else:
            merged.append(current)

    return merged

print(merge_intervals([(1, 3), (2, 6), (8, 10), (10, 12)]), merge_intervals([]), merge_intervals([(5, 7), (1, 2), (2, 4)]))  # Output: [(1, 4), (5, 8)]
```
